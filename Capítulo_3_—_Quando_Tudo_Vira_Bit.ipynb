```python
quantidade = 100000
tamanho_imagem_mb = 5.93

total_mb = quantidade * tamanho_imagem_mb

total_gb = total_mb / 1024

total_tb = total_gb / 1024

print("Total MB:", round(total_mb, 2))
print("Total GB:", round(total_gb, 2))
print("Total TB:", round(total_tb, 2))
```

```python
largura = 1920
altura = 1080

canais = 3
bits_por_canal = 8

bits_por_pixel = canais * bits_por_canal

total_bits = largura * altura * bits_por_pixel

total_bytes = total_bits / 8

megabytes = total_bytes / (1024 ** 2)

print("Tamanho da imagem:", round(megabytes, 2), "MB")
```
